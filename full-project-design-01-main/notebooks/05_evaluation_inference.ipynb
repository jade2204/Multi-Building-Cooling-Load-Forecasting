# 05 — Final evaluation and production inference

Load the champion artifact, review its saved metrics and call the same inference service that the FastAPI app uses.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path[:0] = [str(ROOT / "src"), str(ROOT)]

import json

from cooling_load.api.service import ForecastService
from cooling_load.artifacts import load_bundle
from cooling_load.config import load_config
from cooling_load.io import read_frame

config = load_config(ROOT / "configs/base.yaml")
bundle = load_bundle(config.path("model"))
json.loads(config.path("metrics").read_text())
```

```python
clean = read_frame(config.path("interim_data"))
history = clean[clean.building_id.eq("A")].sort_values("timestamp").tail(200)
service = ForecastService(bundle)
forecast = service.forecast(history.to_dict(orient="records"))
forecast
```
